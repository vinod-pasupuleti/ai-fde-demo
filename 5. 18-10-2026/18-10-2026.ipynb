# Scope of a varible

```python
a = 10

def do_something():
    # local variable a
    a = 0
    a = a+1
    print(a)

do_something()
print(a)
```

```python
a = 10

def do_something():
    global a
    a = a+1
    print(a)

do_something()
print(a)
```

```python
a = 10

def do_something():
    print(a) # assignment and modition is not allowed
    # reading of global variable is allowed inside a function

do_something()
print(a)
```

```python
a = [1,2,3]

def do_something():
    a.append(123) # for collections modification is allowed
    print(a)

do_something()
print(a)
```

### Exception Handling

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

except ZeroDivisionError:
    print("Cannot divide by zero.")

except ValueError:
    print("Invalid input.")

else:
    print("Result:", result)
    print("Calculation successful.")

finally:
    print("Execution completed.")
```

##### Python File Handling + Exception Handling

Let's practise file handling from beginner to intermediate level, combining `open()`, `read()`, `write()`, `append()`, `with`, and exception handling.

### 1. Understanding File Handling

Python supports different file operations:

| Mode | Purpose |
|---|---|
| `r` | Read a file |
| `w` | Write (overwrites existing content) |
| `a` | Append content |
| `x` | Create a new file; error if it exists |
| `r+` | Read and write |
| `rb` | Read binary file |
| `wb` | Write binary file |


```python
#2. Create and Write a File
# Explanation:
#  - open() opens or creates a file.
#  - w creates a new file or overwrites an existing file.
#  - write() writes content.
#  - close() closes the file.

try:
    file = open("sample.txt", "w")

    file.write("Welcome to Python File Handling\n")
    file.write("Learning Python is interesting!")

    file.close()

    print("File created successfully")

except OSError as e:
    print("File operation failed:", e)
```

```python
#3. Read a File
# Read first 10 characters
# file.read(10)

# Read one line
# file.readline()

# Read all lines as a list
# file.readlines()

try:
    file = open("sample.txt", "r")
    content = file.read()
    print(content)
    file.close()
except FileNotFoundError:
    print("File does not exist")
except OSError as e:
    print("Error:", e)

```

```python
#4. Append Content to a File
try:
    with open("sample.txt", "a") as file:
        file.write("\nPractising exception handling.")
    print("Content appended successfully")
except OSError as e:
    print("Unable to append:", e)
```

#### 5. Using `with` Statement (Important)
Instead of manually opening and closing files, use `with`.

Why use `with`?

Python automatically closes the file after leaving the `with` block, even if an exception occurs.

This is the recommended approach in production Python applications.

```python
try:
    with open("sample.txt", "r") as file:
        content = file.read()
        print(content)

except FileNotFoundError:
    print("File not found")

except PermissionError:
    print("Permission denied")

except OSError as e:
    print("File error:", e)
```

#### 6. Read a File Line by Line
Useful when processing large files.

```python
try:
    with open("sample.txt", "r") as file:
        for line_number, line in enumerate(file, start=1):
            # strip() removes leading and trailing whitespace, including newline characters.
            print(f"Line {line_number}: {line.strip()}")
except FileNotFoundError:
    print("File not found")
except OSError as e:
    print("Error:", e)
```

#### 7. Handling Multiple Exceptions
Combine file handling with different exception types.

```python
try:
    filename = input("Enter filename: ")
    with open(filename, "r") as file:
        content = file.read()
    print(content)

except FileNotFoundError:
    print("Error: File does not exist.")

except PermissionError:
    print("Error: You don't have permission.")

except UnicodeDecodeError:
    print("Error: Unable to decode file content.")

except OSError as e:
    print("Operating system error:", e)

except Exception as e:
    print("Unexpected error:", e)

else:
    print("File read successfully.")

finally:
    print("File operation completed.")

# except handles errors.
# else executes only when no exception occurs.
# finally executes regardless of success or failure.
```

#### 8. Practical Example – File Word Counter
create a small real-world utility.

Requirement: Read a text file and calculate:
- Total characters
- Total words
- Total lines

```python
def file_analyzer(filename):

    try:
        with open(filename, "r", encoding="utf-8") as file:

            content = file.read()

            total_characters = len(content)
            total_words = len(content.split())
            total_lines = len(content.splitlines())

            print("File Analysis")
            print("-------------------")
            print("Characters:", total_characters)
            print("Words:", total_words)
            print("Lines:", total_lines)

    except FileNotFoundError:
        print("File not found:", filename)

    except PermissionError:
        print("Permission denied.")

    except UnicodeDecodeError:
        print("Unable to decode file.")

    except OSError as e:
        print("File error:", e)


file_analyzer("sample.txt")
```
